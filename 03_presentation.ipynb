```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot
import numpy as np
```

## Change the global variable to get the dashboard

```python
# Global variables - CHANGE THESE
feature = "Mitochondria Intensity - Outer region Mean"
secondary_feature = "Mitochondria Intensity - Inner region Mean"
category = "Cell Type"
data_path = 'data/Raw_Data_Batch1_LAB2_CTX2.csv'
custom_colors = {
    'PRKN': 'Green',
    'Control': 'Grey',
    'LRRK2': 'Red',
    'SNCA': 'Blue'
}
######
```

```python
main_df = pd.read_csv(data_path)
mitochondria_df = main_df[[ 
                           feature,
                           secondary_feature,
                           'Cellline ID',
                           'Cell Type']]
mitochondria_df = mitochondria_df.dropna()
```

```python
P95 = np.percentile(mitochondria_df[feature], 95)
all_variants_df = mitochondria_df[mitochondria_df[feature] < P95]

control_prkn_df = pd.concat([all_variants_df[all_variants_df['Cell Type'] == 'Control'], all_variants_df[all_variants_df['Cell Type'] == 'PRKN']])

control_df = all_variants_df[all_variants_df['Cell Type'] == 'Control']
prkn_df = all_variants_df[all_variants_df['Cell Type'] == 'PRKN']

my_density_plot = sns.histplot(data=control_prkn_df,
                 x=feature, 
                 hue='Cell Type',
                 stat="percent",
                 common_norm=False,
                 palette=custom_colors,
                 element="step",
                 fill=True,
                 bins=100)

# Plot the Median of the control
control_median = np.percentile(control_df[feature], 50)
prkn_median = np.percentile(prkn_df[feature], 50)

print(control_median)

my_density_plot.axvline(
    x=control_median,
    linewidth=1,
    color='grey',
    ls='dotted'
)

my_density_plot.text(x=control_median + 0.5, 
            y=0.9, 
            s=f"Median: {control_median:.1f}", 
            color='grey', 
            rotation=0,
            transform=my_density_plot.get_xaxis_transform())

# Plot the Median of the prkn

print(prkn_median)

my_density_plot.axvline(
    x=prkn_median,
    linewidth=1,
    color='green',
    ls='dotted'
)

my_density_plot.text(x=prkn_median + 0.5, 
            y=0.8, 
            s=f"Median: {prkn_median:.1f}", 
            color='green', 
            rotation=0,
            transform=my_density_plot.get_xaxis_transform())

my_density_plot.set_title(f"{feature}")

pyplot.show()
```
